```python
import requests
```

```python
## Getting all the studies
## Iterate through every page. When the page doesn't have items, stops

## MGnify returns 219 studies with the keyword 'human gut', we only get 104 when asking their API for this same searching keyword

isMissingPages = True
pageNb = 1
datasets = []
while isMissingPages: 
    url = f"https://www.ebi.ac.uk/metagenomics/api/v2/studies/?search=human%20gut&page={pageNb}"
    print(url)
    response = requests.get(url)
    data = response.json()
    print(len(data['items']))
    if len(data['items']) == 0:
        isMissingPages = False
    else:
        datasets = datasets + data['items']
        print("getting more..")
    pageNb = pageNb + 1
print(f"{len(datasets)} studies found for human gut key word")
```

```python
## Overview of the metadata disponible in the studies

metadataTerms = []
for data in datasets:
    if len(data['metadata']) != {}:
        for key, value in data['metadata'].items():
            print(f"{key} {value}")
            metadataTerms.append({key, value})
print(len(metadataTerms))
```
