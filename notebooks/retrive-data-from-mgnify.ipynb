```python
## The objective of this notebook is to identify the data to use for the catalog from mgnify
```

```python
import requests
```

```python
## Getting all the studies
isMissingPages = True
pageNb = 1
datasets = []
while isMissingPages: 
    url = f"https://www.ebi.ac.uk/metagenomics/api/v2/studies/?search=human%20gut&page={pageNb}"
    response = requests.get(url)
    data = response.json()
    print(data['count'])
    if data['count'] == 0:
        isMissingPages = False
    else:
        datasets.append(data)
        print("getting more..")
    pageNb = pageNb + 1
print(datasets)
```
